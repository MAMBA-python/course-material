## Xlwings Addin Excel


### Opzetten:
- zet in VBA (alt+f11) de Extra/verwijzingen aan op xl_wings
- zet in Excel de optie voor Marcro's van buitenaf vertrouwen aan:
        -  `Enable Trust access to the VBA project object model under File > Options > Trust Center > Trust Center Settings > Macro Settings. `

- Met de grote run knop voer je een functie 'main' uit in het bestand met dezelfde naam.py/in dezelfde map. Deze functie móet main heten.
- Met de UDF (user defined functies) kun je andere functies in je python bestand gebruiken. (Dit bestand kan andere naam heben, maar zelfde map) - Boven de functies in Python moet `@xw.func` staan. Je kunt de functies daarna bij naam aanroepen in een cel. (heel cool!)

#### Nadelen
- Debuggen is moeilijker
- Zorg dat je altijd een kopie van je origineel hebt - je python acties kun je niet ongedaan maken in Excel.
- Je zult moeten werken met functies, maar dat is denk ik een leuke uitdaging, geen bezwaar.
- De documentatie geeft betere voorbeelden dan ChatGPT, dus ik zou die er zeker bijhouden als referentie: https://docs.xlwings.org/en/stable/udfs.html
