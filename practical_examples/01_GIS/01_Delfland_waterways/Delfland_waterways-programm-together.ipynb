```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
from datetime import datetime
import pandas as pd
import xml.etree.ElementTree as ET
import geopandas as gpd
import contextily as ctx
```

### Constants and relative paths to files

```python
# read shapefile with the actual location of the canals
boezem = r"shapefiles/boezemwatergangen.shp"

# read shapefile with the modelled canals
RTCtakken = r"shapefiles/RTC takken voorstel 20160203.shp"

# read csv file with initial water levels
csv_file_time = 'timeseries_0000.csv'

# read an xml file with the model configuration (node coördinates and their names)
csv_file_coords = 'coordinates.csv'
# ns = "{http://www.wldelft.nl/fews}" #standard code, created by the software 


ref_system_xml = 'EPSG:28992'
ref_basemap = 3857
```

```python
df_coords = pd.read_csv(csv_file_coords, index_col=0)

df_coords.head()
```

```python
df_wl = pd.read_csv(csv_file_time)

for col in df_wl.columns:
    if '_HRTC' in col:
        wl = df_wl.loc[0, col]
        col_split = col.split('_')[0]
        df_coords.loc[col_split, 'ini_wl_HRTC'] = wl

df_coords.head()
```

```python
points = gpd.points_from_xy(df_coords.x, df_coords.y, crs=ref_system_xml)
gdf = gpd.GeoDataFrame(df_coords, geometry=points)
```

```python
gdf.head()
gdf = gdf.to_crs(ref_basemap)
```

```python
gdf.explore(column=gdf.ini_wl_HRTC)
```
