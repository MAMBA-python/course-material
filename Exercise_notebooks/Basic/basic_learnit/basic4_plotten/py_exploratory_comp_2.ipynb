### 4. De opties van de plt.plot functie

De functie `plot` kan veel argumenten accepteren. Als je het helpvenster van de functie `plot` bekijkt door `plt.plot(` te typen, krijg je een hulpblokje boven je cursor te zien. Als je `plt.plot?` typt, verschijnt onderaan het notebook een nieuw, scrollbaar venster met de documentatie van `plot`. plt.plot heeft nogal veel opties!

Kort gezegd kan `plot` met één argument worden gebruikt als `plot(y)`. Hiermee worden de waarden van `y` langs de verticale as uitgezet en wordt de horizontale as genummerd vanaf 0. `plot(x, y)` zet `y` uit tegen `x`, en `plot(x, y, formatstring)` zet `y` uit tegen `x` met behulp van de kleuren en markeringen die in `formatstring` zijn gedefinieerd. Dit kan op verschillende manieren worden gebruikt.

Je kunt er bijvoorbeeld de kleur mee definiëren: `'b'` voor blauw, `'r'` voor rood en `'g'` voor groen. Je kunt er ook het lijntype mee definiëren: `'-'` voor een doorgetrokken lijn, `'--'` voor een stippellijn en `':'` voor een lijn met punten. Daarnaast kun je markeringen definiëren, bijvoorbeeld `'o'` voor cirkels en `'s'` voor vierkanten.

Je kunt deze zelfs combineren: `'r--'` geeft een rode stippellijn, terwijl `'go'` groene cirkelvormige markeringen geeft.


Als dat nog niet genoeg is, accepteert plot een groot aantal benoemde argumenten (keyword arguments). Een benoemd argument is een optioneel argument dat aan een functie kan worden toegevoegd. De syntaxis is function(keyword1=value1, keyword2=value2), enzovoort. Om bijvoorbeeld een lijn met een lijndikte van 6 te tekenen (de standaardwaarde is 1), typ je:

```python
plt.plot([1, 2, 3], [2, 4, 3], linewidth=6);
```

Benoemde argumenten (keyword arguments) moeten na de gewone argumenten komen. plot(linewidth=6, [1, 2, 3], [2, 4, 3]) geeft een foutmelding.

Namen kunnen aan de assen worden toegevoegd met de functies `xlabel` en `ylabel`, bijvoorbeeld `plt.xlabel('this is the x-axis')`. Let op dat beide functies een tekenreeks (*string*) als argument gebruiken. Een titel kan aan de figuur worden toegevoegd met het commando `plt.title`. Meerdere krommen kunnen aan dezelfde figuur worden toegevoegd door meerdere plotopdrachten in dezelfde codecel te plaatsen. Ze worden automatisch aan dezelfde figuur toegevoegd.


### Figuur en Figuuropmaak

Telkens wanneer je een plotopdracht in een codecel uitvoert, wordt automatisch een figuur met een standaardformaat aangemaakt. Alle daaropvolgende plotopdrachten in dezelfde codecel worden aan dezelfde figuur toegevoegd.

Als je een ander formaat voor de figuur wilt, kun je eerst een figuur aanmaken met het gewenste formaat met behulp van de syntaxis `plt.figure(figsize=(width, height))`. Alle daaropvolgende plotopdrachten in de codecel worden vervolgens aan deze figuur toegevoegd. Je kunt zelfs een tweede figuur (of een derde, vierde, enzovoort) aanmaken.


```python
plt.figure(figsize=(10, 3))
plt.plot([1, 2, 3], [2, 4, 3], linewidth=6)
plt.title('very wide figure')
plt.figure()  # new figure of default size
plt.plot([1, 2, 3], [1, 3, 1], 'r')
plt.title('second figure');
```

### 6 Codeer style

Programmeurs vinden goede coding style erg belangrijk. Het zorgt ervoor dat code makkelijker te lezen is en dat we fouten veel sneller terug vinden. 

Belangrijke `style` regels zijn:

* spaties rond elk wiskundig symbool (`=`, `+`, `-`, `*`, `/`), maar spaties zijn niet nodig rond `**`
* spaties tussen de argumenten van een functie
* geen spaties rond een gelijkheidsteken bij een benoemd argument (dus `linewidth=3` is correct)
* één spatie na elke komma
* één spatie na elke `#`
* twee spaties vóór een `#` wanneer deze volgt op een Python-opdracht
* geen spatie tussen de functienaam en de lijst met argumenten. `plt.plot(x, y)` is dus een goede stijl, terwijl `plt.plot (x, y)` geen goede stijl is.
* variabelen met kleine letters

Deze regels zijn (een heel klein onderdeel van) de officiële Python-stijlgids, genaamd PEP8. Wanneer deze regels worden toegepast, ziet de code er als volgt uit:


Zie hieronder het verschil in stukjes code - programmeurs vinden het tweede voorbeeld veel prettiger dan het eerste. 


Gebruik vanaf hierafaan alleen nog maar de juiste style bij het typen van code. 

```python
# NIET volgens de style regels


a=1
B=1
c=-6
x=np.linspace(-4,4,100)
y=a*x**2+B*x+c#Compute y for all x values
plt.plot(x,y,linewidth=3)
```

```python
# Wel volgens PEP8
a = 1
b = 1
c = -6
x = np.linspace(-4, 4, 100)
y = a * x**2 + b * x + c  # Compute y for all x values
plt.plot(x, y, linewidth=3)
```

### Kleuren

Als je geen kleur opgeeft voor een plotopdracht, gebruikt `matplotlib` de standaardkleuren. De eerste drie standaardkleuren zijn bijzondere tinten blauw, oranje en groen. De namen van de standaardkleuren bestaan uit een hoofdletter `C`, gevolgd door het nummer, beginnend met nummer `0`. Bijvoorbeeld:


```python
plt.plot([0, 1], [0, 1], 'C0')
plt.plot([0, 1], [1, 2], 'C1')
plt.plot([0, 1], [2, 3], 'C2')
plt.legend(['default blue', 'default orange', 'default green']);
```

Er zijn vijf verschillende manieren om je eigen kleuren op te geven in `matplotlib`-plots. Je kunt hier meer over lezen: [hier](http://matplotlib.org/examples/pylab_examples/color_demo.html). Een handige manier is om de HTML-kleurnamen te gebruiken. De HTML-codes kun je bijvoorbeeld [hier](http://en.wikipedia.org/wiki/Web_colors) vinden.


```python
color1 = 'fuchsia'
color2 = 'lime'
color3 = 'DodgerBlue'
plt.plot([0, 1], [0, 1], color1)
plt.plot([0, 1], [1, 2], color2)
plt.plot([0, 1], [2, 3], color3)
plt.legend([color1, color2, color3]);
```

De coolste (en nerdigste) manier is waarschijnlijk om de xkcd-kleurnamen te gebruiken. Deze moeten worden voorafgegaan door `xkcd:`. De lijst met xkcd-kleurnamen is te vinden op [xkcd](https://xkcd.com/color/rgb/) en bevat favorieten zoals `'baby puke green'` en een aantal bruine kleuren die variëren van `poo` tot `poop brown` en `baby poop brown`. Probeer het uit:

```python
air = np.loadtxt('holland_temperature.dat') 
sea = np.loadtxt('holland_seawater.dat')
plt.subplot(211)
plt.plot(air, 'b', label='air temp')
plt.plot(sea, 'r', label='sea temp')
plt.legend(loc='best')
plt.ylabel('temp (Celcius)')
plt.xlim(0, 11)
plt.xticks([])
plt.subplot(212)
plt.plot(air-sea, 'ko')
plt.xticks(np.linspace(0, 11, 12),
           ['jan','feb','mar','apr','may','jun','jul','aug','sep','oct','nov','dec'])
plt.xlim(0, 11)
plt.ylabel('air - sea temp (Celcius)');
```
